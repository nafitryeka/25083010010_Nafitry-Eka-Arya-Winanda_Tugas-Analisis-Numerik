### Deskripsi Tugas

Tugas ini bertujuan untuk menghitung volume kedua lambung kapal berdasarkan gambar yang diberikan.

Ketentuan yang digunakan dalam perhitungan:

- Satu kotak grid pada gambar bernilai 5 cm + 5 cm = 10 cm.
- Reserve buoyancy tidak dihitung.
- Lantai kapal dianggap datar atau tidak memiliki keel.
- Penampang melintang lambung dianggap berbentuk persegi panjang.
- Volume dihitung menggunakan integrasi numerik metode trapesium.

Volume lambung dihitung berdasarkan perubahan lebar dan kedalaman lambung pada setiap posisi sepanjang kapal.

### Dasar Perhitungan

Karena lebar dan kedalaman lambung berubah terhadap posisi `x`, luas penampang pada setiap posisi dihitung menggunakan:

$$
A(x) = w(x) \times d(x)
$$

dengan:

- $A(x)$ = luas penampang pada posisi $x$ dalam cm²
- $w(x)$ = lebar lambung pada posisi $x$ dalam cm
- $d(x)$ = kedalaman lambung pada posisi $x$ dalam cm

Volume lambung kemudian dihitung menggunakan integral:

$$
V = \int w(x)d(x)\,dx
$$

Karena data yang digunakan berbentuk data diskrit, integral dihitung secara numerik menggunakan metode trapesium.

### Data Pengukuran

Panjang lambung dari sekat buritan sampai sekat haluan adalah sekitar 161,2 cm.


### Perhitungan Luas Penampang

Luas penampang pada setiap titik dihitung dengan:

$$
A(x) = w(x) \times d(x)
$$

Contoh perhitungan pada titik pertama:

$$
A(0) = 17,6 \times 18,7
$$

$$
A(0) = 329,12\text{ cm}^2
$$

Perhitungan dilakukan pada seluruh titik pengukuran.


### Integrasi Numerik Metode Trapesium

Metode trapesium digunakan untuk menghitung luas di bawah kurva luas penampang terhadap posisi `x`.

Rumus metode trapesium:

$$
\int_a^b f(x)\,dx
\approx
\sum_{i=0}^{n-1}
\frac{f(x_i)+f(x_{i+1})}{2}
(x_{i+1}-x_i)
$$

Dalam tugas ini:

$$
f(x)=w(x)d(x)
$$

sehingga:

$$
V \approx
\sum_{i=0}^{n-1}
\frac{A_i+A_{i+1}}{2}
(x_{i+1}-x_i)
$$

Hasil integrasi digunakan untuk mendapatkan volume satu lambung dalam satuan cm³.

### Implementasi Python

```python
import numpy as np

# Skala gambar
# 1 kotak = 5 cm + 5 cm = 10 cm
skala_kotak = 10

print("Skala 1 kotak =", skala_kotak, "cm")


# Data pengukuran
x = np.array([
    0.0,
    16.0,
    32.2,
    48.3,
    64.4,
    80.5,
    96.7,
    112.7,
    128.9,
    145.0,
    161.2
])

w = np.array([
    17.6,
    21.0,
    21.1,
    21.4,
    21.4,
    21.4,
    21.4,
    21.2,
    20.5,
    17.7,
    13.2
])

d = np.array([
    18.7,
    24.2,
    25.9,
    25.9,
    25.9,
    25.9,
    25.9,
    25.9,
    25.9,
    25.9,
    25.9
])


# Menghitung luas penampang
# A(x) = w(x) × d(x)

A = w * d

print("\nLuas penampang:")
print("x (cm)    w (cm)    d (cm)    A (cm²)")
print("-" * 45)

for i in range(len(x)):
    print(
        f"{x[i]:6.1f}    "
        f"{w[i]:6.1f}    "
        f"{d[i]:6.1f}    "
        f"{A[i]:8.2f}"
    )


# Menghitung selisih x
# Δx = x(i+1) - x(i)

dx = np.diff(x)


# Metode trapesium
#
# V_i = ((A_i + A_i+1) / 2) × Δx

volume_interval = []

print("\nPerhitungan metode trapesium:")
print()

for i in range(len(dx)):
    luas_rata = (A[i] + A[i + 1]) / 2
    volume = luas_rata * dx[i]

    volume_interval.append(volume)

    print(f"Interval {i + 1}")
    print(f"Δx = {x[i + 1]:.1f} - {x[i]:.1f} = {dx[i]:.1f} cm")
    print(
        f"V{i + 1} = "
        f"(({A[i]:.2f} + {A[i + 1]:.2f}) / 2) × {dx[i]:.1f}"
    )
    print(f"V{i + 1} = {volume:.2f} cm³")
    print()

volume_interval = np.array(volume_interval)


# Menjumlahkan seluruh volume interval

volume_satu_cm3 = np.sum(volume_interval)

print("Jumlah volume satu lambung:")
print()

for i, volume in enumerate(volume_interval):
    if i == 0:
        print(f"{volume:.2f}", end="")
    else:
        print(f" + {volume:.2f}", end="")

print(f" = {volume_satu_cm3:.2f} cm³")


# Konversi satuan

volume_satu_m3 = volume_satu_cm3 / 1_000_000
volume_satu_liter = volume_satu_cm3 / 1_000

print("\nKonversi volume satu lambung:")
print(f"{volume_satu_cm3:.2f} cm³")
print(f"= {volume_satu_cm3:.2f} / 1.000.000")
print(f"= {volume_satu_m3:.5f} m³")
print(f"= {volume_satu_liter:.2f} liter")


# Volume kedua lambung

volume_kedua_cm3 = volume_satu_cm3 * 2
volume_kedua_m3 = volume_kedua_cm3 / 1_000_000
volume_kedua_liter = volume_kedua_cm3 / 1_000

print("\nVolume kedua lambung:")
print()
print("V_total = 2 × V_satu_lambung")
print(
    f"V_total = 2 × {volume_satu_cm3:.2f}"
)
print(
    f"V_total = {volume_kedua_cm3:.2f} cm³"
)
print(
    f"V_total = {volume_kedua_m3:.5f} m³"
)
print(
    f"V_total = {volume_kedua_liter:.2f} liter"
)


# Hasil akhir

print("\nHasil akhir:")
print(f"Volume satu lambung  = {volume_satu_m3:.5f} m³")
print(f"Volume kedua lambung = {volume_kedua_m3:.5f} m³")
```

### Interpretasi

Berdasarkan hasil perhitungan menggunakan metode integrasi numerik trapesium, volume satu lambung kapal diperoleh sekitar 0,083 m³ atau 82,95 liter.

Karena kapal memiliki dua lambung yang dianggap identik, volume total kedua lambung adalah sekitar 0,166 m³ atau 165,90 liter.

Metode trapesium digunakan karena lebar dan kedalaman lambung tidak konstan sepanjang kapal. Dengan menggunakan beberapa titik pengukuran, luas penampang pada setiap posisi dapat dihitung dan kemudian diintegralkan secara numerik untuk memperoleh pendekatan volume lambung.
